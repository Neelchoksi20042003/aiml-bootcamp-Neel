# Task 02 — Plot a Function Against Its Derivative

## 1. Function

Given:

f(x) = x³ - 3x

Using the power rule:

f'(x) = 3x² - 3

Set the derivative equal to zero:

3x² - 3 = 0
3x² = 3
x² = 1

Therefore:

x = -1, 1


## 2. Identifying the Peak and Valley

At x = -1:

f(-1) = (-1)³ - 3(-1)
      = -1 + 3
      = 2

So the point is (-1, 2).

The derivative changes from positive to negative around x = -1,
so this is a local maximum (peak).


At x = 1:

f(1) = 1³ - 3(1)
     = 1 - 3
     = -2

So the point is (1, -2).

The derivative changes from negative to positive around x = 1,
so this is a local minimum (valley).

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-3, 3, 200)

f = x**3 - 3*x
df = 3*x**2 - 3

critical_x = np.array([-1, 1])
critical_y = critical_x**3 - 3*critical_x
```

```python
fig, axes = plt.subplots(
    2, 1,
    figsize=(7, 7),
    sharex=True
)

# Original function
axes[0].plot(x, f)
axes[0].scatter(critical_x, critical_y, s=70)

axes[0].axhline(0, linewidth=0.8)
axes[0].axvline(0, linewidth=0.8)

axes[0].set_ylabel("f(x)")
axes[0].set_title("f(x) = x³ - 3x")
axes[0].grid(True)


# Derivative
axes[1].plot(x, df)
axes[1].scatter(critical_x, [0, 0], s=70)

axes[1].axhline(0, linewidth=0.8)
axes[1].axvline(0, linewidth=0.8)

axes[1].set_xlabel("x")
axes[1].set_ylabel("f'(x)")
axes[1].set_title("f'(x) = 3x² - 3")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

## 3. Interpretation

The derivative is positive when the original function is rising
and negative when the original function is falling.

At x = -1, the derivative changes from positive to negative,
so the original function has a local maximum (peak).

At x = 1, the derivative changes from negative to positive,
so the original function has a local minimum (valley).

Therefore:

Peak: (-1, 2)
Valley: (1, -2)

## 4. Questions

### Where the derivative is positive, is the function rising or falling?

When the derivative is positive, the function is rising.

### A cubic can have two flat spots. What is the difference between them?

One can be a local maximum (peak) and the other can be a local
minimum (valley). The difference is determined by the change in
sign of the derivative.

At x = -1:

positive → negative

Therefore it is a peak.

At x = 1:

negative → positive

Therefore it is a valley.

### What does it mean when the derivative touches zero but does not cross it?

A zero derivative does not automatically mean there is a peak or
valley. If the derivative does not change sign, the point may simply
be a stationary/flat point.

## 5. Conclusion

The function has a peak where the derivative crosses zero from positive
to negative, and a valley where it crosses zero from negative to positive.
