```python
# TASK 04 — THE CHAIN RULE

import sympy as sp
import numpy as np

# 1. Numerical derivative

def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

# 2. FIRST FUNCTION
# f(x) = (3x + 1)^5

x = sp.symbols('x')

f1 = (3*x + 1)**5

df1_sympy = sp.diff(f1, x)

print("=" * 60)
print("FIRST FUNCTION")
print("=" * 60)

print("Function:")
print("f(x) = (3x + 1)^5")

print("\nOuter function:")
print("u^5")

print("\nInner function:")
print("u = 3x + 1")

print("\nDerivative by hand:")
print("df/dx = 5(3x + 1)^4 × 3")
print("df/dx = 15(3x + 1)^4")

print("\nSymPy derivative:")
print(df1_sympy)

# 3. First function — numerical verification

def f1_func(x):
    return (3*x + 1)**5


def f1_exact(x):
    return 15*(3*x + 1)**4


print("\n" + "=" * 60)
print("FIRST FUNCTION — NUMERICAL VERIFICATION")
print("=" * 60)

for value in [0, 1, 2]:

    exact = f1_exact(value)

    numerical = numerical_derivative(
        f1_func,
        value
    )

    difference = abs(exact - numerical)

    print(f"\nx = {value}")
    print("Exact derivative    :", exact)
    print("Numerical derivative :", numerical)
    print("Difference           :", difference)

# 4. THREE-LAYER CHAIN
# f(x) = sin((2x + 1)^2)

f2 = sp.sin((2*x + 1)**2)

df2_sympy = sp.diff(f2, x)

print("\n\n" + "=" * 60)
print("THREE-LAYER CHAIN")
print("=" * 60)

print("Function:")
print("f(x) = sin((2x + 1)^2)")

print("\nLayers from outer to inner:")
print("1. sin(u)")
print("2. u^2")
print("3. 2x + 1")

print("\nDerivative by hand:")

print("Outer derivative:")
print("cos((2x + 1)^2)")

print("\nMiddle derivative:")
print("2(2x + 1)")

print("\nInner derivative:")
print("2")

print("\nMultiply:")
print("cos((2x + 1)^2) × 2(2x + 1) × 2")

print("\nFinal derivative:")
print("4(2x + 1)cos((2x + 1)^2)")

print("\nSymPy derivative:")
print(df2_sympy)

# 5. Three-layer numerical verification

def f2_func(x):
    return np.sin((2*x + 1)**2)


def f2_exact(x):
    return 4*(2*x + 1)*np.cos((2*x + 1)**2)


print("\n" + "=" * 60)
print("THREE-LAYER — NUMERICAL VERIFICATION")
print("=" * 60)

for value in [0, 0.5, 1]:

    exact = f2_exact(value)

    numerical = numerical_derivative(
        f2_func,
        value
    )

    difference = abs(exact - numerical)

    print(f"\nx = {value}")
    print("Exact derivative    :", exact)
    print("Numerical derivative :", numerical)
    print("Difference           :", difference)
```

Q1. For a chain of three functions, how many derivatives get multiplied?

Answer:

For a chain of three functions, three derivatives are multiplied together. Each derivative represents how much one layer changes in response to the output of the previous layer.

Mathematically:

$$ f(g(h(x))) $$

has:

$$ \boxed{ f'(g(h(x)))\times g'(h(x))\times h'(x) } $$
Q2. What are the three layers in \(\sin((2x+1)^2)\)?

From outside to inside:

$$ \boxed{\sin(u)} $$ $$ \boxed{u=v^2} $$ $$ \boxed{v=2x+1} $$

In simple terms:

The three layers are the sine function, the square operation, and the inner linear function \(2x+1\).

Q3. What happens if one link has a very small derivative?

Suppose:

$$ \frac{df}{dx} = A\times B\times C $$

and:

$$ C=0.001 $$

Then the final derivative becomes small because we're multiplying by \(0.001\).

So:

If one derivative in a chain is very small, the overall derivative can become very small because all the derivatives are multiplied together.

This is important in deep learning because repeatedly multiplying small derivatives can lead to a vanishing gradient.

The chain rule multiplies derivatives because the change in one function is passed through the next function, so we need to combine the rate of change at every layer. Multiplying the derivatives gives us the total effect of a small change in the original input on the final output.
