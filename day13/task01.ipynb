```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp


# ==========================================
# 1. SymPy setup
# ==========================================

x = sp.symbols('x')


# ==========================================
# 2. Define symbolic functions
# ==========================================

f = x**4
g = 3*x**2 + 2*x - 7
h = 1/x


# ==========================================
# 3. Compute exact derivatives using SymPy
# ==========================================

f_derivative = sp.diff(f, x)
g_derivative = sp.diff(g, x)
h_derivative = sp.diff(h, x)

print("f(x) =", f)
print("f'(x) =", f_derivative)

print("\ng(x) =", g)
print("g'(x) =", g_derivative)

print("\nh(x) =", h)
print("h'(x) =", h_derivative)


# ==========================================
# 4. Evaluate derivatives at x = 2
# ==========================================

print("\n--- Derivatives at x = 2 ---")

print("f'(2) =", f_derivative.subs(x, 2))
print("g'(2) =", g_derivative.subs(x, 2))
print("h'(2) =", h_derivative.subs(x, 2))


# ==========================================
# 5. Numerical derivative function
# ==========================================

def numerical_derivative(func, x, h=1e-5):
    return (func(x + h) - func(x - h)) / (2 * h)


# ==========================================
# 6. Normal Python versions of functions
# ==========================================

def f_func(x):
    return x**4


def g_func(x):
    return 3*x**2 + 2*x - 7


def h_func(x):
    return 1/x


# ==========================================
# 7. Compare numerical and exact derivatives
# ==========================================

print("\n--- Numerical vs Exact at x = 2 ---")

numerical_f = numerical_derivative(f_func, 2)
numerical_g = numerical_derivative(g_func, 2)
numerical_h = numerical_derivative(h_func, 2)

exact_f = float(f_derivative.subs(x, 2))
exact_g = float(g_derivative.subs(x, 2))
exact_h = float(h_derivative.subs(x, 2))

print("f'(2)")
print("Exact:", exact_f)
print("Numerical:", numerical_f)

print("\ng'(2)")
print("Exact:", exact_g)
print("Numerical:", numerical_g)

print("\nh'(2)")
print("Exact:", exact_h)
print("Numerical:", numerical_h)


# ==========================================
# 8. Find where f'(x) = 0
# ==========================================

critical_points = sp.solve(sp.Eq(f_derivative, 0), x)

print("\n--- Critical Points ---")
print("f'(x) =", f_derivative)
print("Where f'(x) = 0:", critical_points)


# ==========================================
# 9. Plot f(x)
# ==========================================

x_values = np.linspace(-2, 2, 400)
y_values = x_values**4

plt.figure(figsize=(8, 5))

plt.plot(x_values, y_values)

plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("f(x) = x⁴")

plt.grid(True)
plt.show()
```

ans:1 The power rule says that when differentiating \(x^n\), bring the exponent \(n\) in front as a multiplier and reduce the exponent by 1.

ans2: It uses points on both sides of \(x\), making the approximation symmetric and generally more accurate than the forward difference.

ans3:The approximation becomes less local and can have significant approximation error because the function may curve over the larger interval.

ans4: Floating-point precision and cancellation errors can dominate, causing the numerical derivative to become inaccurate.
