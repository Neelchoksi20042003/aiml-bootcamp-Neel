```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 4 * np.pi, 200)

y_sin = np.sin(x)
y_cos = np.cos(x)

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(x, y_sin, label="sin(x)")
ax.plot(x, y_cos, label="cos(x)")


ax.set_title("Sine and Cosine Functions")
ax.set_xlabel("x (radians)")
ax.set_ylabel("Function value")


ax.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt




np.random.seed(42)

x = np.random.uniform(0, 10, 100)
noise = np.random.normal(0, 2, 100)

y = 2 * x + noise

colour = np.random.uniform(0, 100, 100)

size = np.random.uniform(20, 300, 100)

fig, ax = plt.subplots(figsize=(8, 5))

scatter = ax.scatter(
    x,
    y,
    c=colour,
    s=size,
    alpha=0.7
)
ax.set_title("Relationship Between X and Y")


ax.set_xlabel("X Value (units)")


ax.set_ylabel("Y Value (units)")


fig.colorbar(
    scatter,
    ax=ax,
    label="Third Variable (units)"
)

# Display the graph
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


np.random.seed(42)


x_line = np.linspace(0, 10, 100)
y_line = np.sin(x_line)


x_scatter = np.random.uniform(0, 10, 50)
y_scatter = 2 * x_scatter + np.random.normal(0, 2, 50)


categories = ["A", "B", "C", "D"]
values = [25, 40, 30, 50]

scores = np.random.normal(70, 10, 500)


fig, axes = plt.subplots(
    2,
    2,
    figsize=(11, 8)
)


axes[0, 0].plot(
    x_line,
    y_line,
    label="sin(x)"
)

axes[0, 0].set_title("Line Plot")
axes[0, 0].set_xlabel("X value (units)")
axes[0, 0].set_ylabel("sin(x) value")
axes[0, 0].legend()


axes[0, 1].scatter(
    x_scatter,
    y_scatter,
    alpha=0.7,
    label="Data points"
)

axes[0, 1].set_title("Scatter Plot")
axes[0, 1].set_xlabel("X value (units)")
axes[0, 1].set_ylabel("Y value (units)")
axes[0, 1].legend()


axes[1, 0].bar(
    categories,
    values
)

axes[1, 0].set_title("Bar Chart")
axes[1, 0].set_xlabel("Category")
axes[1, 0].set_ylabel("Value (units)")

axes[1, 1].hist(
    scores,
    bins=30
)

axes[1, 1].set_title("Histogram")
axes[1, 1].set_xlabel("Score (marks)")
axes[1, 1].set_ylabel("Frequency")


fig.suptitle("Four Common Data Visualisation Types")


plt.tight_layout()

plt.show()
```

```python
np.random.seed(42)

it_salary = np.random.normal(70000, 10000, 100)
hr_salary = np.random.normal(55000, 8000, 100)
finance_salary = np.random.normal(65000, 9000, 100)
sales_salary = np.random.normal(60000, 12000, 100)

salary_data = [
    it_salary,
    hr_salary,
    finance_salary,
    sales_salary
]

departments = [
    "IT",
    "HR",
    "Finance",
    "Sales"
]

fig, ax = plt.subplots(figsize=(9, 5))

ax.boxplot(
    salary_data,
    tick_labels=departments
)

ax.set_title("Salary Distribution by Department")
ax.set_xlabel("Department")
ax.set_ylabel("Salary (₹ per year)")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


rng = np.random.default_rng(42)

n_students = 200

study_hours = rng.uniform(1, 10, n_students)

attendance = rng.uniform(60, 100, n_students)

exam_scores = (
    35
    + 4 * study_hours
    + 0.35 * attendance
    + rng.normal(0, 8, n_students)
)

exam_scores = np.clip(exam_scores, 0, 100)

passed = exam_scores >= 60


average_score_passed = exam_scores[passed].mean()
average_score_failed = exam_scores[~passed].mean()

average_attendance_passed = attendance[passed].mean()
average_attendance_failed = attendance[~passed].mean()

categories = ["Passed", "Failed"]


fig, axes = plt.subplots(
    2,
    2,
    figsize=(12, 9)
)

fig.suptitle(
    "How Do Study Time and Attendance Relate to Student Performance?",
    fontsize=16
)



axes[0, 0].scatter(
    study_hours,
    exam_scores,
    alpha=0.6,
    label="Students"
)

axes[0, 0].set_title("Study Hours vs Exam Score")
axes[0, 0].set_xlabel("Study hours (hours/week)")
axes[0, 0].set_ylabel("Exam score (marks)")
axes[0, 0].legend()



axes[0, 1].hist(
    exam_scores,
    bins=20,
    alpha=0.7,
    label="Students"
)

axes[0, 1].axvline(
    exam_scores.mean(),
    linestyle="--",
    label="Mean score"
)

axes[0, 1].set_title("Distribution of Exam Scores")
axes[0, 1].set_xlabel("Exam score (marks)")
axes[0, 1].set_ylabel("Number of students (students)")
axes[0, 1].legend()


average_scores = [
    average_score_passed,
    average_score_failed
]

axes[1, 0].bar(
    categories,
    average_scores
)

axes[1, 0].set_title("Average Exam Score by Result")
axes[1, 0].set_xlabel("Result")
axes[1, 0].set_ylabel("Average score (marks)")


average_attendance = [
    average_attendance_passed,
    average_attendance_failed
]

axes[1, 1].bar(
    categories,
    average_attendance
)

axes[1, 1].set_title("Average Attendance by Result")
axes[1, 1].set_xlabel("Result")
axes[1, 1].set_ylabel("Attendance (%)")



plt.tight_layout()

fig.savefig(
    "report_figure.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

The figure shows the relationship between study hours, attendance, and student exam performance across 200 students. The main takeaway is that exam scores generally increase with study time, while students who pass also tend to have higher average attendance
