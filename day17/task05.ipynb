```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(42)

# --------------------------------------------------
# 1. TYPE I ERROR SIMULATION
# --------------------------------------------------

count = 0
n_tests = 1000

for _ in range(n_tests):
    # Same distribution → no real difference
    a = rng.normal(60, 12, 100)
    b = rng.normal(60, 12, 100)

    p_value = stats.ttest_ind(a, b).pvalue

    if p_value < 0.05:
        count += 1

false_positive_rate = count / n_tests

print("TYPE I ERROR")
print("-" * 30)
print("Significant results:", count)
print("False-positive rate:", round(false_positive_rate, 3))
print("Percentage:", round(false_positive_rate * 100, 2), "%")

# --------------------------------------------------
# 2. SIGNIFICANCE VS PRACTICAL IMPORTANCE
# --------------------------------------------------

n = 5000

group_a = rng.normal(60, 10, n)
group_b = rng.normal(60.2, 10, n)

t_stat, p_value = stats.ttest_ind(group_a, group_b)

mean_a = np.mean(group_a)
mean_b = np.mean(group_b)
difference = mean_b - mean_a

print("\nSIGNIFICANCE VS PRACTICAL IMPORTANCE")
print("-" * 40)
print("Group A mean:", round(mean_a, 3))
print("Group B mean:", round(mean_b, 3))
print("Mean difference:", round(difference, 3))
print("t-statistic:", round(t_stat, 3))
print("p-value:", p_value)

if p_value < 0.05:
    print("\nResult: Statistically significant")
else:
    print("\nResult: Not statistically significant")

print("Practical difference: Only about 0.2 points")
```

## 1. Type I Error

A Type I error occurs when we reject the null hypothesis even though it is actually true.

In this simulation, both groups are generated from the same distribution, so there is truly no difference between them. We run an independent t-test 1000 times and count how often the p-value is below 0.05.

If the test is working correctly, about 5% of the tests should incorrectly appear significant.

**Lesson:** Running many tests and reporting only significant results is dangerous because some false-positive results are expected to appear by chance.

If we run 20 independent tests at α = 0.05, we expect approximately:

20 × 0.05 = 1 false-positive result.

---

## 2. Statistical Significance vs Practical Importance

Next, we create two very large groups whose means differ by only 0.2 points.

With thousands of observations, even this tiny difference can produce a statistically significant p-value.

**Statistical significance** tells us whether an observed difference is unlikely to be due to random chance.

**Practical importance** asks whether the size of that difference is large enough to matter in the real world.

In our example, a 0.2-point difference can be statistically significant because the sample size is very large. However, such a small difference may be practically unimportant and may not justify taking action.
