```python
# TASK 02 — 95% CONFIDENCE INTERVAL

import numpy as np
import pandas as pd
from scipy import stats

# Load the cleaned dataset
df = pd.read_csv("student_performance.csv")

# Remove missing values
df = df.dropna()

# Remove the impossible 99-hour outlier
df = df[df["study_hours"] != 99]

# Select final scores
final = df["final_score"]

# 1. MEAN AND STANDARD ERROR

n = len(final)
mean = final.mean()

# Sample standard deviation
sd = final.std(ddof=1)

# Standard error = standard deviation / sqrt(sample size)
se = sd / np.sqrt(n)

print("FULL DATASET")
print("-" * 40)
print(f"Sample size (n): {n}")
print(f"Mean: {mean:.2f}")
print(f"Standard deviation: {sd:.2f}")
print(f"Standard error: {se:.4f}")

# 2. 95% CONFIDENCE INTERVAL

lo, hi = stats.t.interval(
    0.95,
    df=n - 1,
    loc=mean,
    scale=se
)

print(f"95% CI: ({lo:.2f}, {hi:.2f})")
print(f"\nWe are 95% confident that the true average final score")
print(f"lies between {lo:.2f} and {hi:.2f}.")

# 3. SEE HOW SAMPLE SIZE CHANGES THE INTERVAL

def confidence_interval(data, sample_size):

    # Take a sample of the required size
    sample = data.iloc[:sample_size]

    n = len(sample)
    mean = sample.mean()
    sd = sample.std(ddof=1)
    se = sd / np.sqrt(n)

    lo, hi = stats.t.interval(
        0.95,
        df=n - 1,
        loc=mean,
        scale=se
    )

    width = hi - lo

    return mean, lo, hi, width


# Calculate for 30 students
mean_30, lo_30, hi_30, width_30 = confidence_interval(
    final,
    30
)

# Calculate for 2000 students
mean_2000, lo_2000, hi_2000, width_2000 = confidence_interval(
    final,
    2000
)

print("\nSAMPLE SIZE COMPARISON")
print("-" * 60)

print(
    f"30 students   → CI: ({lo_30:.2f}, {hi_30:.2f}) "
    f"| Width: {width_30:.2f}"
)

print(
    f"2000 students → CI: ({lo_2000:.2f}, {hi_2000:.2f}) "
    f"| Width: {width_2000:.2f}"
)

print("\nLarger sample → narrower confidence interval.")

# 4. OPTIONAL: COMPARE WIDTH VISUALLY

sample_sizes = [30, 2000]
widths = [width_30, width_2000]

print("\nWIDTH COMPARISON")
print("-" * 40)

for size, width in zip(sample_sizes, widths):
    print(f"{size} students: CI width = {width:.2f}")
```

## Lookup Questions & Answers

### 1. Why does a bigger sample give a narrower interval?

Because standard error decreases as sample size increases:

`SE = s / √n`

So the confidence interval becomes narrower.

### 2. Does a 95% CI mean there's a 95% chance the true value is in this one interval?

No.

It means that if we repeatedly took samples and calculated 95% confidence intervals, about 95% of those intervals would contain the true population mean.

### 3. What would a 99% confidence interval look like compared to a 95% one?

A **99% confidence interval is wider** than a 95% confidence interval because higher confidence requires a larger range of possible values.
