```python
# TASK 04 — CHI-SQUARE TEST

import numpy as np
import pandas as pd
from scipy import stats

# Load dataset
df = pd.read_csv("student_performance.csv")

# Clean data
df = df.dropna()
df = df[df["study_hours"] != 99]

# 1. CREATE CATEGORICAL VARIABLES

# High attendance = 1, Low attendance = 0
cat1 = (df["attendance_pct"] >= 80).astype(int)

# Passed = 1, Failed = 0
# Change the condition if your dataset has a different
# definition of passing.
passed = (df["final_score"] >= 40).astype(int)

# 2. CREATE 2x2 CONTINGENCY TABLE

table = np.zeros((2, 2), dtype=int)

for attendance_group, pass_status in zip(cat1, passed):
    table[attendance_group, pass_status] += 1

print("CONTINGENCY TABLE")
print("-" * 45)

print("Rows: Attendance (0 = Low, 1 = High)")
print("Columns: Result (0 = Fail, 1 = Pass)")
print()

print(table)

# 3. CHI-SQUARE TEST

chi2, p_value, dof, expected = stats.chi2_contingency(table)

print("\nCHI-SQUARE TEST RESULTS")
print("-" * 45)

print(f"Chi-square statistic : {chi2:.4f}")
print(f"p-value              : {p_value:.6f}")
print(f"Degrees of freedom   : {dof}")

print("\nEXPECTED FREQUENCIES")
print(expected)

# 4. HYPOTHESIS

print("\nNULL HYPOTHESIS (H₀)")
print("High/low attendance and pass/fail status are independent.")

# 5. DECISION

alpha = 0.05

print("\nDECISION")
print("-" * 45)

if p_value < alpha:
    print("Reject H₀.")
    print("There is a statistically significant relationship between")
    print("attendance category and pass/fail status.")
else:
    print("Fail to reject H₀.")
    print("There is not enough evidence of a relationship between")
    print("attendance category and pass/fail status.")

# 6. PLAIN-ENGLISH CONCLUSION

print("\nPLAIN-ENGLISH CONCLUSION")
print("-" * 45)

if p_value < alpha:
    print(
        f"The chi-square test gives p = {p_value:.4f}, which is below "
        f"0.05. Therefore, attendance category and pass/fail status "
        f"appear to be related in this dataset."
    )
else:
    print(
        f"The chi-square test gives p = {p_value:.4f}, which is above "
        f"0.05. Therefore, this dataset does not provide sufficient "
        f"evidence that attendance category and pass/fail status are related."
    )
```

# TASK 04 — A CHI-SQUARE TEST: ARE TWO CATEGORIES RELATED?

## Required Information

- Select two categorical variables:
  - High/low attendance
  - Pass/fail status
- Create a **2 × 2 contingency table**.
- **Null hypothesis (H₀):** The two categorical variables are independent (unrelated).
- Run `stats.chi2_contingency()`.
- Report:
  - Chi-square statistic
  - p-value
  - Degrees of freedom
  - Expected frequencies
- Use significance level `α = 0.05`.
- If `p < 0.05` → reject H₀ → evidence of a relationship.
- If `p >= 0.05` → fail to reject H₀ → insufficient evidence of a relationship.
- Write the conclusion in plain English.

## Lookup Questions & Answers

### 1. Why can't you use a t-test here instead of chi-square?

A t-test is generally used to compare the means of a numerical variable between groups.

A chi-square test is designed to determine whether two **categorical variables** are associated.

Here we are comparing categories such as attendance (`high/low`) and result (`pass/fail`), so the chi-square test is appropriate.

### 2. What does the contingency table's `expected` output represent?

Expected frequencies are the counts we would expect in each cell of the contingency table **if the two categorical variables were independent**.

They are compared with the observed frequencies to calculate the chi-square statistic.

### 3. How does this connect to the idea of independence from the probability session?

Two events are independent when knowing one does not change the probability of the other.

For example, if attendance and passing were independent, knowing whether a student had high attendance would not change their probability of passing.

The chi-square test checks whether the observed data provides evidence against this independence.
