```python
# TASK 03 — INDEPENDENT-SAMPLES T-TEST

import numpy as np
import pandas as pd
from scipy import stats

# Load dataset
df = pd.read_csv("student_performance.csv")

# Clean data
df = df.dropna()
df = df[df["study_hours"] != 99]

# Select final score and attendance
final = df["final_score"]
attendance = df["attendance_pct"]

# 1. SPLIT INTO TWO GROUPS

# High attendance: 80% or more
groupA = final[attendance >= 80]

# Low attendance: below 80%
groupB = final[attendance < 80]

# 2. NULL HYPOTHESIS

print("NULL HYPOTHESIS (H₀):")
print("The two groups have the same true average final score.")

# 3. CALCULATE GROUP MEANS

mean_A = groupA.mean()
mean_B = groupB.mean()

difference = mean_A - mean_B

print("\nGROUP RESULTS")
print("-" * 45)
print(f"High Attendance Mean : {mean_A:.2f}")
print(f"Low Attendance Mean  : {mean_B:.2f}")
print(f"Difference            : {difference:.2f}")

# 4. INDEPENDENT-SAMPLES T-TEST

t_stat, p_value = stats.ttest_ind(
    groupA,
    groupB
)

print("\nT-TEST RESULTS")
print("-" * 45)
print(f"t-statistic : {t_stat:.4f}")
print(f"p-value     : {p_value:.6f}")

# 5. SIGNIFICANCE DECISION

alpha = 0.05

if p_value < alpha:
    print("\nDecision: Statistically significant.")
    print("Reject the null hypothesis.")
else:
    print("\nDecision: Not statistically significant.")
    print("Fail to reject the null hypothesis.")

# 6. PRACTICAL SIGNIFICANCE

print("\nPRACTICAL INTERPRETATION")
print("-" * 45)

print(
    f"The difference in average final scores is "
    f"{abs(difference):.2f} points."
)

if abs(difference) < 2:
    print("The difference is relatively small in practical terms.")
elif abs(difference) < 5:
    print("The difference may have moderate practical importance.")
else:
    print("The difference is large enough to potentially matter in practice.")

# 7. PLAIN-ENGLISH CONCLUSION

print("\nPLAIN-ENGLISH CONCLUSION")
print("-" * 45)

if p_value < alpha:
    print(
        f"Students with high attendance scored an average of "
        f"{mean_A:.2f}, compared with {mean_B:.2f} for students "
        f"with low attendance. The difference of {abs(difference):.2f} "
        f"points is statistically significant (p = {p_value:.4f})."
    )
else:
    print(
        f"Students with high attendance scored an average of "
        f"{mean_A:.2f}, compared with {mean_B:.2f} for students "
        f"with low attendance. The difference of {abs(difference):.2f} "
        f"points is not statistically significant (p = {p_value:.4f})."
    )
```

# TASK 03 — A T-TEST: DO TWO GROUPS DIFFER?

## Required Information

- Split students into two groups:
  - High attendance: `attendance >= 80`
  - Low attendance: `attendance < 80`
- **Null hypothesis (H₀):** The two groups have the same true average final score.
- Calculate the mean of both groups.
- Calculate the difference between the group means.
- Run an independent-samples t-test using `stats.ttest_ind()`.
- Report the **t-statistic** and **p-value**.
- Use significance level: `α = 0.05`.
- If `p < 0.05` → statistically significant → reject H₀.
- If `p >= 0.05` → not statistically significant → fail to reject H₀.
- Also consider **practical significance**: whether the difference in average scores is large enough to matter in practice.

## Lookup Questions & Answers

### 1. What exactly does the null hypothesis say, in your own words?

The null hypothesis says that there is **no real difference between the average final scores of the two groups**. Any observed difference is assumed to be due to random variation.

### 2. If p = 0.03, what does that number literally mean?

It means that, **assuming the null hypothesis is true**, there is a 3% probability of observing a difference at least as extreme as the one found in the sample.

Since `0.03 < 0.05`, the result is statistically significant at the 5% level.

### 3. Your difference is significant — but is it large enough to act on? How do you decide?

Statistical significance does not automatically mean practical importance.

Check the **size of the difference between the group means** and consider whether that difference is meaningful in the real-world context.

A very small difference can be statistically significant with a large sample, while a larger difference may not be statistically significant with a small sample.
