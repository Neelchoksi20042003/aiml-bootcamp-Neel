```python
import numpy as np

# TASK 04 — A Second Bayes Scenario
# Scenario: Spam Filter

# One reproducible random generator
rng = np.random.default_rng(0)

# 1. Define the scenario
# We want to answer:

# P(Spam | Email contains "free")#
# Given:

# P(Spam) = 0.40
# P("free" | Spam) = 0.30
# P("free" | Ham) = 0.02

P_spam = 0.40

# Probability that a spam email contains "free"
P_free_given_spam = 0.30

# Probability that a normal email contains "free"
# This is the false-positive rate.
P_free_given_ham = 0.02

P_ham = 1 - P_spam


# 2. Calculate P("free")
# An email can contain "free" in two ways:
# 1. It is spam AND contains "free"
# 2. It is ham AND contains "free"
# P(free) =
# P(free | spam)P(spam)
# +
# P(free | ham)P(ham)

P_free = (
    P_free_given_spam * P_spam
    +
    P_free_given_ham * P_ham
)


# 3. Apply Bayes' theorem
# P(Spam | free)
# =
# P(free | Spam)P(Spam)
# P(free)

P_spam_given_free = (
    P_free_given_spam * P_spam
    / P_free
)

# 4. Display all terms

print("=" * 60)
print("TASK 04 — BAYES' THEOREM: SPAM FILTER")
print("=" * 60)
print("\nPrior:")
print("P(Spam) =", P_spam)

print("\nProbability of Ham:")
print("P(Ham) =", P_ham)

print("\nLikelihood:")
print('P("free" | Spam) =', P_free_given_spam)

print("\nFalse-positive rate:")
print('P("free" | Ham) =', P_free_given_ham)

print("\nEvidence:")
print('P("free") =', P_free)

print("\nPosterior:")
print('P(Spam | "free") =', P_spam_given_free)

print(
    f'\nIf an email contains "free", '
    f'the probability that it is spam is '
    f'{P_spam_given_free * 100:.2f}%'
)


# 5. Monte Carlo Simulation#
# Simulate 1,000,000 emails.
#
# Step 1:
# Randomly decide whether each email is spam.
#
# Step 2:
# If spam, 30% contain "free".
#
# If ham, 2% contain "free".
#
# Step 3:
# Among emails containing "free", calculate
# how many are actually spam.

n = 1_000_000

# Randomly assign spam status
spam = rng.random(n) < P_spam

# Generate random values for whether "free" appears
random_values = rng.random(n)

# Create an array for whether the word "free" appears
contains_free = np.zeros(n, dtype=bool)

# Spam emails contain "free" with probability 30%
contains_free[spam] = (
    random_values[spam] < P_free_given_spam
)

# Ham emails contain "free" with probability 2%
ham = ~spam

contains_free[ham] = (
    random_values[ham] < P_free_given_ham
)


# 6. Calculate simulated posterior

emails_with_free = contains_free.sum()

spam_with_free = np.sum(
    spam & contains_free
)

P_spam_given_free_simulated = (
    spam_with_free / emails_with_free
)


# 7. Compare simulation with Bayes calculation

print("\n" + "=" * 60)
print("MONTE CARLO SIMULATION")
print("=" * 60)

print("Total emails simulated:", n)

print("Total spam emails:", spam.sum())

print('Emails containing "free":', emails_with_free)

print(
    'Spam emails containing "free":',
    spam_with_free
)

print(
    '\nSimulated P(Spam | "free") =',
    P_spam_given_free_simulated
)

print(
    'Theoretical P(Spam | "free") =',
    P_spam_given_free
)

print(
    "Difference =",
    abs(
        P_spam_given_free_simulated
        - P_spam_given_free
    )
)

print(
    "\nSimulation agrees with theory?",
    np.isclose(
        P_spam_given_free_simulated,
        P_spam_given_free,
        atol=0.01
    )
)

# 8. Final Bayes lesson

print("\n" + "=" * 60)
print("LESSON")
print("=" * 60)

print(
    'The prior was P(Spam) = 40%, representing how common spam '
    'was before observing the word "free".'
)

print(
    'Seeing "free" acts as evidence and updates the prior into '
    f'a posterior probability of about '
    f'{P_spam_given_free * 100:.2f}% that the email is spam.'
)
```
