```python
import numpy as np
from itertools import product

# ============================================================
# TASK 02 — Conditional Probability and Independence
# ============================================================

# Use the same RNG seed/setup from Task 01
rng = np.random.default_rng(0)


# ------------------------------------------------------------
# 1. Create the sample space for two dice
# ------------------------------------------------------------
# Each outcome is an ordered pair:
# (first die, second die)
#
# There are 6 × 6 = 36 possible outcomes.

outcomes = list(product(range(1, 7), repeat=2))

print("Number of possible outcomes:", len(outcomes))
print("Sample space:")
print(outcomes)


# ------------------------------------------------------------
# 2. Define the events
# ------------------------------------------------------------

# Event A: sum of the two dice is 8
A = [outcome for outcome in outcomes if sum(outcome) == 8]

# Event B: first die is 3
B = [outcome for outcome in outcomes if outcome[0] == 3]

# A AND B: sum is 8 AND first die is 3
A_and_B = [outcome for outcome in outcomes
           if sum(outcome) == 8 and outcome[0] == 3]


print("\nEvent A (sum = 8):")
print(A)

print("\nEvent B (first die = 3):")
print(B)

print("\nEvent A AND B:")
print(A_and_B)


# ------------------------------------------------------------
# 3. Calculate P(sum = 8)
# ------------------------------------------------------------

P_A = len(A) / len(outcomes)

print("\nP(sum = 8) =", P_A)


# ------------------------------------------------------------
# 4. Calculate P(first die = 3)
# ------------------------------------------------------------

P_B = len(B) / len(outcomes)

print("P(first die = 3) =", P_B)


# ------------------------------------------------------------
# 5. Conditional probability using the formula
#
# P(A | B) = P(A AND B) / P(B)
# ------------------------------------------------------------

P_A_and_B = len(A_and_B) / len(outcomes)

P_A_given_B_formula = P_A_and_B / P_B

print("\nP(A AND B) =", P_A_and_B)

print("P(A | B) using formula =", P_A_given_B_formula)


# ------------------------------------------------------------
# 6. Conditional probability by counting
#
# Once we know the first die is 3, we restrict our world
# to only outcomes where the first die is 3.
#
# B = [(3,1), (3,2), (3,3), (3,4), (3,5), (3,6)]
#
# Only (3,5) has sum = 8.
#
# Therefore:
# P(A | B) = 1 / 6
# ------------------------------------------------------------

restricted_world = B

favorable_in_restricted_world = [
    outcome for outcome in restricted_world
    if sum(outcome) == 8
]

P_A_given_B_counting = (
    len(favorable_in_restricted_world)
    / len(restricted_world)
)

print("\nRestricted world:", restricted_world)

print(
    "Favorable outcomes inside restricted world:",
    favorable_in_restricted_world
)

print(
    "P(A | B) by direct counting =",
    P_A_given_B_counting
)


# ------------------------------------------------------------
# 7. Check that both methods agree
# ------------------------------------------------------------

print(
    "\nDo both conditional probability methods agree?",
    np.isclose(P_A_given_B_formula, P_A_given_B_counting)
)


# ------------------------------------------------------------
# 8. Test whether A and B are independent
#
# Independent events satisfy:
#
# P(A AND B) = P(A) × P(B)
# ------------------------------------------------------------

product_probability = P_A * P_B

print("\nIndependence test for:")
print("A = sum is 8")
print("B = first die is 3")

print("P(A AND B) =", P_A_and_B)
print("P(A) × P(B) =", product_probability)

print(
    "Are A and B independent?",
    np.isclose(P_A_and_B, product_probability)
)


# ------------------------------------------------------------
# 9. Find an independent pair of events
#
# C = first die is even
# D = second die is even
#
# Since the dice are independent, knowing the first die
# does not change the probability of the second die being even.
# ------------------------------------------------------------

C = [outcome for outcome in outcomes if outcome[0] % 2 == 0]

D = [outcome for outcome in outcomes if outcome[1] % 2 == 0]

C_and_D = [
    outcome for outcome in outcomes
    if outcome[0] % 2 == 0 and outcome[1] % 2 == 0
]

P_C = len(C) / len(outcomes)
P_D = len(D) / len(outcomes)
P_C_and_D = len(C_and_D) / len(outcomes)

print("\n--- Independent Pair ---")

print("C = first die is even")
print("D = second die is even")

print("P(C) =", P_C)
print("P(D) =", P_D)
print("P(C AND D) =", P_C_and_D)
print("P(C) × P(D) =", P_C * P_D)

print(
    "Independent?",
    np.isclose(P_C_and_D, P_C * P_D)
)

# ------------------------------------------------------------
# 10. Dependent pair
#
# E = sum is 8
# F = first die is 3
# ------------------------------------------------------------

E = [outcome for outcome in outcomes if sum(outcome) == 8]

F = [outcome for outcome in outcomes if outcome[0] == 3]

E_and_F = [
    outcome for outcome in outcomes
    if sum(outcome) == 8 and outcome[0] == 3
]

P_E = len(E) / len(outcomes)
P_F = len(F) / len(outcomes)
P_E_and_F = len(E_and_F) / len(outcomes)

print("\n--- Dependent Pair ---")

print("E = sum is 8")
print("F = first die is 3")

print("P(E) =", P_E)
print("P(F) =", P_F)
print("P(E AND F) =", P_E_and_F)
print("P(E) × P(F) =", P_E * P_F)

print(
    "Independent?",
    np.isclose(P_E_and_F, P_E * P_F)
)

# ------------------------------------------------------------
# 11. Monte Carlo simulation
# ------------------------------------------------------------
# Generate 100,000 two-dice rolls.
#
# We first filter to the condition:
# first die = 3
#
# Then among those rolls, calculate the fraction
# where the total sum is 8.
# ------------------------------------------------------------

n = 100_000

d1 = rng.integers(1, 7, size=n)
d2 = rng.integers(1, 7, size=n)

# Condition B: first die is 3
condition = (d1 == 3)

# Keep only rolls where first die is 3
filtered_d1 = d1[condition]
filtered_d2 = d2[condition]

# Among the filtered rolls, check sum = 8
sum_is_8 = (filtered_d1 + filtered_d2 == 8)

P_A_given_B_simulated = sum_is_8.mean()

print("\n--- Monte Carlo Verification ---")

print("Number of simulated rolls:", n)
print("Rolls where first die = 3:", condition.sum())

print(
    "Simulated P(sum = 8 | first die = 3) =",
    P_A_given_B_simulated
)

print(
    "Theoretical P(sum = 8 | first die = 3) =",
    P_A_given_B_counting
)

print(
    "Simulation agrees with theory?",
    np.isclose(
        P_A_given_B_simulated,
        P_A_given_B_counting,
        atol=0.01
    )
)
```
