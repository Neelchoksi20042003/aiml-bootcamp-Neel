```python
import numpy as np

# ============================================================
# TASK 03 — The Medical Test Problem
# ============================================================

# Use one reproducible random generator
rng = np.random.default_rng(0)


# ------------------------------------------------------------
# 1. Given probabilities
# ------------------------------------------------------------

# Disease affects 2% of the population
P_disease = 0.02

# Sensitivity:
# Probability of a positive test when the person is sick
P_positive_given_disease = 0.98

# Specificity:
# Probability of a negative test when the person is healthy
P_negative_given_healthy = 0.90

# Therefore, false positive rate:
# P(positive | healthy) = 1 - specificity
P_positive_given_healthy = 1 - P_negative_given_healthy


# ------------------------------------------------------------
# 2. Bayes' Theorem
#
# P(Disease | Positive)
# =
# P(Positive | Disease) * P(Disease)
# ---------------------------------
# P(Positive)
#
# First calculate P(Positive)
# ------------------------------------------------------------

P_healthy = 1 - P_disease

P_positive = (
    P_positive_given_disease * P_disease
    +
    P_positive_given_healthy * P_healthy
)

P_disease_given_positive = (
    P_positive_given_disease * P_disease
    / P_positive
)


# ------------------------------------------------------------
# 3. Display every term
# ------------------------------------------------------------

print("=" * 60)
print("BAYES' THEOREM — MEDICAL TEST")
print("=" * 60)

print("\nPrior probability:")
print("P(Disease) =", P_disease)

print("\nSensitivity:")
print("P(Positive | Disease) =", P_positive_given_disease)

print("\nSpecificity:")
print("P(Negative | Healthy) =", P_negative_given_healthy)

print("\nFalse Positive Rate:")
print("P(Positive | Healthy) =", P_positive_given_healthy)

print("\nProbability of being healthy:")
print("P(Healthy) =", P_healthy)

print("\nProbability of a positive test:")
print("P(Positive) =", P_positive)

print("\nBayes' Theorem:")
print("P(Disease | Positive) =",
      P_disease_given_positive)

print(
    f"\nPercentage of positive tests that are actually disease cases: "
    f"{P_disease_given_positive * 100:.2f}%"
)


# ------------------------------------------------------------
# 4. Explain the surprising result
# ------------------------------------------------------------

print("\n--- Why is the result surprising? ---")

print(
    "Even though the test is highly sensitive, the disease is rare."
)

print(
    "The large healthy population produces many false positives, "
    "which can outnumber the true positives from the small sick population."
)


# ------------------------------------------------------------
# 5. Monte Carlo simulation — 1,000,000 people
# ------------------------------------------------------------

n = 1_000_000

# Randomly assign disease status:
# True  = sick
# False = healthy
disease = rng.random(n) < P_disease


# Simulate test results.
#
# For sick people:
# 98% should test positive.
#
# For healthy people:
# 10% should test positive.

test_positive = np.zeros(n, dtype=bool)

# Generate random values for each person
random_values = rng.random(n)

# Sick people test positive with probability 0.98
test_positive[disease] = (
    random_values[disease] < P_positive_given_disease
)

# Healthy people test positive with probability 0.10
healthy = ~disease

test_positive[healthy] = (
    random_values[healthy] < P_positive_given_healthy
)


# ------------------------------------------------------------
# 6. Calculate simulated P(Disease | Positive)
# ------------------------------------------------------------

positive_people = test_positive.sum()

sick_and_positive = np.sum(
    disease & test_positive
)

P_disease_given_positive_simulated = (
    sick_and_positive / positive_people
)


print("\n" + "=" * 60)
print("MONTE CARLO SIMULATION")
print("=" * 60)

print("Total people simulated:", n)

print("Actual sick people:", disease.sum())

print("Positive tests:", positive_people)

print("Positive tests that are actually sick:",
      sick_and_positive)

print(
    "\nSimulated P(Disease | Positive) =",
    P_disease_given_positive_simulated
)

print(
    "Theoretical P(Disease | Positive) =",
    P_disease_given_positive
)

print(
    "Difference =",
    abs(
        P_disease_given_positive_simulated
        - P_disease_given_positive
    )
)


# ------------------------------------------------------------
# 7. Prevalence experiment
#
# Recalculate P(Disease | Positive) for:
# 0.1%, 2%, and 20% prevalence.
# ------------------------------------------------------------

prevalences = [0.001, 0.02, 0.20]

print("\n" + "=" * 60)
print("PREVALENCE TABLE")
print("=" * 60)

print(
    f"{'Prevalence':<15}"
    f"{'P(Positive)':<20}"
    f"{'P(Disease | Positive)':<25}"
)

print("-" * 60)

for prevalence in prevalences:

    healthy_probability = 1 - prevalence

    # Total probability of getting a positive test
    positive_probability = (
        P_positive_given_disease * prevalence
        +
        P_positive_given_healthy * healthy_probability
    )

    # Bayes' theorem
    disease_given_positive = (
        P_positive_given_disease * prevalence
        / positive_probability
    )

    print(
        f"{prevalence * 100:<15.1f}"
        f"{positive_probability:<20.4f}"
        f"{disease_given_positive * 100:<25.2f}%"
    )


# ------------------------------------------------------------
# 8. Final lesson
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("LESSON")
print("=" * 60)

print(
    "A test can be highly sensitive and specific but still have "
    "many false positives when the disease is very rare."
)

print(
    "Therefore, the probability that a positive result is truly "
    "positive depends not only on test performance but also on "
    "the prevalence of the condition."
)
```
