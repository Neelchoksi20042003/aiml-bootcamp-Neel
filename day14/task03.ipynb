# TASK 03 — Sweep the Learning Rate Until It Breaks

In this task, I will investigate how the learning rate affects
gradient descent.

I will minimize:

f(x) = (x - 5)^2

starting from:

x = 0

The gradient is:

f'(x) = 2(x - 5)

I will test several learning rates and observe three behaviours:

1. Too small — very slow convergence
2. Suitable — fast convergence
3. Too large — divergence

## Learning Rate

The gradient descent update rule is:

x_new = x - learning_rate * gradient

The learning rate controls how far we move in the direction
suggested by the gradient.

A small learning rate produces small updates.

A suitable learning rate moves toward the minimum efficiently.

A learning rate that is too large can repeatedly overshoot
the minimum and eventually diverge.

```python
import numpy as np
import matplotlib.pyplot as plt

# Reusable Gradient Descent Function

def gradient_descent(grad_fn, start, lr, steps):
    x = start
    history = [x]

    for _ in range(steps):
        x = x - lr * grad_fn(x)
        history.append(x)

    return history

# Function and Gradient

def f(x):
    return (x - 5) ** 2


def grad(x):
    return 2 * (x - 5)


# TASK 03
# Learning Rate Sweep
learning_rates = [
    0.001,
    0.01,
    0.1,
    0.5,
    0.9,
    1.0,
    1.05
]

steps = 60
start = 0.0

print("LEARNING RATE SWEEP")
print("=" * 50)

results = {}

for lr in learning_rates:

    path = np.array(
        gradient_descent(
            grad_fn=grad,
            start=start,
            lr=lr,
            steps=steps
        )
    )

    results[lr] = path

    print(
        f"Learning rate = {lr:<5} "
        f"-> final x = {path[-1]:.6f}"
    )

# First six positions for selected learning rates

print("\nFIRST SIX POSITIONS")
print("=" * 50)

for lr in [0.001, 0.1, 1.05]:

    path = results[lr]

    print(f"\nLearning rate = {lr}")

    for i, x in enumerate(path[:6]):
        print(f"Step {i}: x = {x:.6f}")

# Three Side-by-Side Descent Paths

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

selected_rates = [
    (0.001, "Too Small — Crawls"),
    (0.1, "Good — Converges"),
    (1.05, "Too Large — Diverges")
]

for ax, (lr, title) in zip(axes, selected_rates):

    path = results[lr]

    # Function curve
    x_values = np.linspace(-2, 12, 400)
    y_values = f(x_values)

    ax.plot(
        x_values,
        y_values,
        label="f(x) = (x - 5)^2"
    )

    # Descent path
    ax.plot(
        path,
        f(path),
        marker="o",
        markersize=3,
        linestyle="--",
        label=f"lr = {lr}"
    )

    ax.set_xlabel("x")
    ax.set_ylabel("f(x)")
    ax.set_title(title)
    ax.grid(True)
    ax.legend()

plt.suptitle("Effect of Learning Rate on Gradient Descent")
plt.tight_layout()
plt.show()

# Divergence Demonstration

print("\nDIVERGING RUN — lr = 1.05")
print("=" * 50)

diverging_path = results[1.05]

for i, x in enumerate(diverging_path[:10]):
    print(f"Step {i}: x = {x:.6f}")


# Find the Divergence Threshold

print("\nDIVERGENCE THRESHOLD")
print("=" * 50)

# For f(x) = (x - 5)^2:
#
# x_new = x - lr * 2(x - 5)
#
# Let e = x - 5
#
# e_new = (1 - 2*lr)e
#
# For convergence:
#
# |1 - 2*lr| < 1
#
# This gives:
#
# 0 < lr < 1
#
# Therefore the critical learning rate is 1.0.

threshold = 1.0

print(f"Critical learning rate = {threshold}")
print("For lr < 1.0: convergence")
print("For lr = 1.0: reaches the minimum in one step")
print("For lr > 1.0: divergence")

# Verify experimentally around the threshold

print("\nEXPERIMENTAL CHECK")
print("=" * 50)

test_rates = [0.99, 1.0, 1.01]

for lr in test_rates:

    path = np.array(
        gradient_descent(
            grad_fn=grad,
            start=0.0,
            lr=lr,
            steps=10
        )
    )

    print(
        f"lr = {lr} -> "
        f"x after 10 steps = {path[-1]:.6f}"
    )

# Final Summary

print("\nFINAL SUMMARY")
print("=" * 50)

print("Function: f(x) = (x - 5)^2")
print("Starting point: x = 0")
print("Minimum: x = 5")
print("Divergence threshold: lr = 1.0")
print()
print("A learning rate below 1.0 converges.")
print("A learning rate of 1.0 reaches x = 5 immediately.")
print("A learning rate above 1.0 diverges.")
```

## Results and Observations

The learning-rate sweep shows three different behaviours.

### lr = 0.001

The optimizer moves toward x = 5 extremely slowly.

The learning rate is so small that each update is tiny.

### lr = 0.01

The optimizer converges, but still requires many iterations.

### lr = 0.1

The optimizer converges quickly and smoothly toward x = 5.

### lr = 0.5

The optimizer also converges quickly.

### lr = 0.9

The optimizer converges, but the updates can oscillate around the
minimum before settling.

### lr = 1.0

The optimizer reaches x = 5 in exactly one update from x = 0.

### lr = 1.05

The optimizer diverges because each update moves farther away
from the minimum.

## Finding the Divergence Threshold

For:

f(x) = (x - 5)^2

the gradient is:

gradient = 2(x - 5)

The update is:

x_new = x - lr * 2(x - 5)

Consider the distance from the minimum:

e = x - 5

Then:

e_new = (1 - 2lr)e

For convergence, the distance from the minimum must become smaller:

|1 - 2lr| < 1

Solving this gives:

0 < lr < 1

Therefore, the critical learning rate is:

lr = 1.0

At learning rates greater than 1.0, every step makes the distance
from the minimum larger.

## Why Does a Too-Large Learning Rate Diverge?

When the learning rate is too large, the update overshoots the minimum
and lands on the other side farther away from it.

The next gradient points back toward the minimum, but the oversized
update overshoots again by an even larger distance, causing the
iterations to move farther and farther away.

## Why Is a Very Small Learning Rate a Problem?

A very small learning rate may eventually reach the minimum, but it
can require a huge number of iterations.

In real machine-learning problems, this means more computation,
longer training time, and slower experimentation.

Therefore, we want a learning rate that provides fast progress while
remaining stable.

## Can We Change the Learning Rate During Training?

Yes.

The learning rate does not always have to remain fixed.

A common idea is to use a larger learning rate during the early
stages of training for faster progress and gradually reduce it later
for more stable convergence.

This is called a learning-rate schedule.

The general idea is:

large learning rate → fast early progress

smaller learning rate → stable final convergence
