````python
# TASK 05 — Hunt the Outliers

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")


# ============================================================
# 1. Boxplot of SalePrice
# ============================================================

plt.figure(figsize=(10, 4))

sns.boxplot(x=df["SalePrice"])

plt.title("SalePrice — Boxplot")
plt.xlabel("Sale Price")
plt.show()


# ============================================================
# 2. IQR Method — Detect SalePrice Outliers
# ============================================================

q1 = df["SalePrice"].quantile(0.25)
q3 = df["SalePrice"].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

price_outliers = df[
    (df["SalePrice"] < lower_bound) |
    (df["SalePrice"] > upper_bound)
]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("\nNumber of SalePrice outliers:", len(price_outliers))

print("\nSalePrice outliers:")
display(
    price_outliers[
        ["Id", "SalePrice", "GrLivArea", "OverallQual", "Neighborhood"]
    ].sort_values("SalePrice")
)


# ============================================================
# 3. Houses Breaking the Size-Price Relationship
# ============================================================

# Very large living area but surprisingly low SalePrice
size_price_outliers = df[
    (df["GrLivArea"] > 4000) &
    (df["SalePrice"] < 200000)
]

print("\nNumber of size-price outliers:", len(size_price_outliers))

print("\nLarge houses with low SalePrice:")
display(
    size_price_outliers[
        ["Id", "GrLivArea", "SalePrice", "OverallQual", "Neighborhood"]
    ].sort_values("GrLivArea", ascending=False)
)


# ============================================================
# 4. Visualize Size-Price Outliers
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="GrLivArea",
    y="SalePrice",
    alpha=0.5
)

# Highlight the size-price outliers
sns.scatterplot(
    data=size_price_outliers,
    x="GrLivArea",
    y="SalePrice",
    color="red",
    s=100,
    label="Size-Price Outliers"
)

plt.title("GrLivArea vs SalePrice — Size-Price Outliers")
plt.xlabel("Above Ground Living Area (sq ft)")
plt.ylabel("Sale Price")
plt.legend()
plt.show()


# ============================================================
# 5. Print Individual Judgement Information
# ============================================================

print("\nOUTLIER JUDGEMENT")
print("-" * 60)

for _, row in size_price_outliers.iterrows():
    print(
        f"Id {row['Id']}: "
        f"GrLivArea={row['GrLivArea']}, "
        f"SalePrice={row['SalePrice']} "
        f"→ Investigate as a potential genuine rarity or data issue."
    )
````

# TASK 05 — Hunt the Outliers

An **outlier** is an observation that is unusually far from the rest of the data.

## 1. SalePrice Boxplot

The boxplot shows the distribution of SalePrice and highlights unusually low or high prices as points outside the whiskers.

## 2. IQR Method

The IQR method uses:

- Q1 = 25th percentile
- Q3 = 75th percentile
- IQR = Q3 − Q1

### Outlier rule

A value is flagged as an outlier if:

- Value < Q1 − 1.5 × IQR
- OR
- Value > Q3 + 1.5 × IQR

The code calculates these boundaries and lists all SalePrice outliers.

## 3. Size-Price Outliers

We look for houses satisfying:

- GrLivArea > 4000 sq ft
- SalePrice < $200,000

These houses are unusual because very large living areas would generally be expected to have higher prices.

The code automatically calculates how many such houses exist and displays them.

## 4. Outlier Judgement

### SalePrice outliers

A statistical outlier is **not automatically an error**.

Possible explanations include:
- Luxury/high-value property
- Unusual property characteristics
- Very poor property condition
- Data-entry problem

**Action:** Investigate the original record before deleting it.

### Large GrLivArea + Low SalePrice

These observations break the normal size-price relationship.

They could be:
- Genuine unusual houses
- Houses with poor condition or other unusual characteristics
- Data-entry errors

**Action:** Check the original data and relevant variables before deciding whether to remove them.

## Look It Up — Answers

### Why can one outlier badly distort a model?

Extreme values can strongly influence statistics such as the mean and correlation and can pull a regression model toward the unusual observation.

### Should you always delete outliers?

**No.**

Keep an outlier when it represents a real and meaningful observation.

Remove or correct it when there is evidence that it is a data-entry or measurement error.

### Univariate outlier vs relationship outlier

**Univariate outlier:**
An unusual value in one variable.

Example:
A house with an extremely high SalePrice.

**Relationship outlier:**
A point that does not follow the normal relationship between two variables.

Example:
A very large house with an unexpectedly low SalePrice.

### Final takeaway

Outlier detection should be used to **investigate unusual observations, not automatically delete them**.

The IQR method identifies statistically unusual SalePrice values, while the GrLivArea/SalePrice analysis identifies houses that behave unusually relative to the size-price relationship.

# TASK 06 — Plotnine / Grammar of Graphics

## What is Grammar of Graphics?

Grammar of Graphics is a way of constructing charts using separate components or layers:

1. **Data** → the dataframe
2. **Aesthetics (`aes`)** → map columns to x, y, colour, etc.
3. **Geoms** → decide how the data is displayed, such as points, bars, or boxplots
4. **Labels/Theme** → customize the final chart

Plotnine follows this layered approach and has syntax similar to R's ggplot2.

## Why do some people prefer the ggplot style?

The layered structure makes complex charts easier to build step-by-step. 
Each component has a clear purpose, so additional layers such as trend lines, facets, or themes can be added without rewriting the whole chart.

## Plotnine vs Seaborn vs Plotly

### Seaborn
Best for:
- Quick statistical EDA
- Correlation plots
- Boxplots
- Scatterplots
- Fast exploration

### Plotnine
Best for:
- Grammar-of-graphics style
- Layered and structured charts
- Users who like ggplot2-style syntax
- Reusable, declarative visualizations

### Plotly
Best for:
- Interactive charts
- Hover information
- Interactive filtering/zooming
- Dashboards and web applications

## Plotnine vs Seaborn — My comparison

I found **Seaborn more natural for quick EDA** because the syntax is straightforward and charts can be created quickly.

Plotnine feels more structured because the chart is built explicitly in layers: data → aesthetics → geometry → labels/theme.
