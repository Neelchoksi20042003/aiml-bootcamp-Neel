## Task 01: Find Eigenvectors by Hand, Then Verify with NumPy

Matrix:

A = [[4, 1],
     [2, 3]]

The eigenvalue equation is:

Av = λv

To find the eigenvalues, we solve:

det(A - λI) = 0

## Part 1 — Find Eigenvalues by Hand

Given:

A = [[4, 1],
     [2, 3]]

The identity matrix is:

I = [[1, 0],
     [0, 1]]

Therefore:

A - λI = [[4 - λ, 1],
           [2, 3 - λ]]

For an eigenvalue:

det(A - λI) = 0

So:

(4 - λ)(3 - λ) - (1)(2) = 0

Expanding:

12 - 4λ - 3λ + λ² - 2 = 0

λ² - 7λ + 10 = 0

Factorising:

(λ - 5)(λ - 2) = 0

Therefore:

λ₁ = 5
λ₂ = 2

## Part 2 — Find Eigenvector for λ = 5

We use:

(A - λI)v = 0

For λ = 5:

A - 5I = [[-1, 1],
           [ 2,-2]]

Let:

v = [x, y]

Then:

[-1, 1] [x] = [0]
[ 2,-2] [y]   [0]

From the first equation:

-x + y = 0

Therefore:

y = x

Choose x = 1.

Then:

y = 1

Therefore, an eigenvector is:

v₁ = [1, 1]

## Part 3 — Find Eigenvector for λ = 2

For λ = 2:

A - 2I = [[2, 1],
           [2, 1]]

Again:

(A - λI)v = 0

Let:

v = [x, y]

Then:

2x + y = 0

Therefore:

y = -2x

Choose x = 1.

Then:

y = -2

Therefore, an eigenvector is:

v₂ = [1, -2]

## Hand-Work Summary

The eigenvalues are:

λ₁ = 5
λ₂ = 2

Corresponding eigenvectors are:

For λ₁ = 5:

v₁ = [1, 1]

For λ₂ = 2:

v₂ = [1, -2]

These eigenvectors represent directions. Any non-zero scalar multiple
of an eigenvector represents the same eigenvector direction.

```python
import numpy as np

# Create the matrix A
A = np.array([
    [4., 1.],
    [2., 3.]
])

# Calculate eigenvalues and eigenvectors
vals, vecs = np.linalg.eig(A)

print("Matrix A:")
print(A)

print("\nEigenvalues:")
print(vals)

print("\nEigenvectors (columns):")
print(vecs)
```

## Part 4 — Compare Hand-Work with NumPy

NumPy returns the eigenvalues approximately as:

[5, 2]

These match the eigenvalues calculated by hand.

NumPy returns eigenvectors as columns of the `vecs` matrix.

The eigenvectors may look different from the hand answers because
NumPy normalises them to have length 1.

For example:

[1, 1]

and

[0.7071, 0.7071]

represent the same direction.

```python
# Eigenvectors calculated by hand
v1_hand = np.array([1., 1.])
v2_hand = np.array([1., -2.])

# Normalize the hand eigenvectors
v1_normalized = v1_hand / np.linalg.norm(v1_hand)
v2_normalized = v2_hand / np.linalg.norm(v2_hand)

print("Normalized hand eigenvector 1:")
print(v1_normalized)

print("\nNormalized hand eigenvector 2:")
print(v2_normalized)

print("\nNumPy eigenvector 1:")
print(vecs[:, 0])

print("\nNumPy eigenvector 2:")
print(vecs[:, 1])
```

```python
# Compare the directions of hand and NumPy eigenvectors

comparison_1 = abs(np.dot(v1_normalized, vecs[:, 0]))
comparison_2 = abs(np.dot(v2_normalized, vecs[:, 1]))

print("Direction comparison for eigenvector 1:")
print(comparison_1)

print("\nDirection comparison for eigenvector 2:")
print(comparison_2)

print("\nDo eigenvector 1 directions match?")
print(np.isclose(comparison_1, 1.0))

print("\nDo eigenvector 2 directions match?")
print(np.isclose(comparison_2, 1.0))
```

```python
# Verify Av = λv for each eigenvalue/eigenvector pair

for i in range(len(vals)):
    eigenvalue = vals[i]
    eigenvector = vecs[:, i]

    left_side = A @ eigenvector
    right_side = eigenvalue * eigenvector

    print(f"\nEigenvalue {i + 1}: {eigenvalue}")
    print("Eigenvector:")
    print(eigenvector)

    print("\nA @ v:")
    print(left_side)

    print("\nλ * v:")
    print(right_side)

    print("\nDoes Av = λv?")
    print(np.allclose(left_side, right_side))
```

## Part 5 — Conclusion

The eigenvalues calculated by hand were:

λ₁ = 5
λ₂ = 2

The corresponding eigenvectors were:

v₁ = [1, 1]
v₂ = [1, -2]

NumPy confirmed the same eigenvalues.

The NumPy eigenvectors were normalised to length 1, so they differed
from the hand-computed vectors only by a scale factor (and possibly
a sign).

The definition of an eigenvector was also verified directly:

Av = λv

for both eigenvalue/eigenvector pairs.

### Key concepts learned

1. Eigenvalues are found using:

   det(A - λI) = 0

2. Eigenvectors are found using:

   (A - λI)v = 0

3. The fundamental definition is:

   Av = λv

4. Eigenvectors represent directions.

5. Any non-zero scalar multiple of an eigenvector is also an
   eigenvector.

6. NumPy returns eigenvectors as columns.

7. `np.linalg.eig()` performs the numerical eigenvalue/eigenvector
   calculation.
