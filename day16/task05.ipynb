````python
# TASK 05 — FIT A DISTRIBUTION TO REALISTIC DATA

import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Fixed seed for reproducibility
rng = np.random.default_rng(0)

# 1. CREATE REALISTIC DATA

# Pretend we don't know the original distribution.
# We generate waiting-time data from an exponential distribution.
data = rng.exponential(scale=2, size=10_000)

# 2. PLOT THE RAW DATA

plt.figure(figsize=(10, 6))

plt.hist(
    data,
    bins=50,
    density=True,
    alpha=0.6
)

plt.title("Histogram of Data")
plt.xlabel("Value")
plt.ylabel("Density")

plt.show()

# 3. FIT TWO CANDIDATE DISTRIBUTIONS

# Fit a Normal distribution.
mu, sigma = stats.norm.fit(data)

# Fit an Exponential distribution.
loc, scale = stats.expon.fit(data)


# Print fitted parameters
print("FITTED PARAMETERS")
print("-" * 50)

print("Normal Distribution")
print(f"Mean (μ) = {mu:.4f}")
print(f"Standard Deviation (σ) = {sigma:.4f}")

print("\nExponential Distribution")
print(f"Location = {loc:.4f}")
print(f"Scale = {scale:.4f}")

# 4. CREATE FITTED CURVES

# Create x-values covering the data range.
x = np.linspace(data.min(), data.max(), 1000)

# Normal fitted PDF
normal_pdf = stats.norm.pdf(
    x,
    loc=mu,
    scale=sigma
)

# Exponential fitted PDF
exponential_pdf = stats.expon.pdf(
    x,
    loc=loc,
    scale=scale
)

# 5. OVERLAY FITTED CURVES ON HISTOGRAM

plt.figure(figsize=(10, 6))

# Histogram of original data
plt.hist(
    data,
    bins=50,
    density=True,
    alpha=0.5,
    label="Data"
)

# Normal fitted curve
plt.plot(
    x,
    normal_pdf,
    linewidth=2,
    label="Fitted Normal"
)

# Exponential fitted curve
plt.plot(
    x,
    exponential_pdf,
    linewidth=2,
    label="Fitted Exponential"
)

plt.title("Distribution Fitting")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()

plt.show()

# 6. CALCULATE LOG-LIKELIHOOD

# Log-likelihood measures how well a distribution explains
# the observed data.
#
# Higher log-likelihood = better fit.

ll_norm = np.sum(
    stats.norm.logpdf(
        data,
        loc=mu,
        scale=sigma
    )
)

ll_expon = np.sum(
    stats.expon.logpdf(
        data,
        loc=loc,
        scale=scale
    )
)

# 7. COMPARE THE FITS

print("\nLOG-LIKELIHOOD COMPARISON")
print("-" * 50)

print(f"Normal Log-Likelihood      : {ll_norm:.2f}")
print(f"Exponential Log-Likelihood : {ll_expon:.2f}")

# 8. IDENTIFY THE BEST FIT

if ll_norm > ll_expon:
    best_fit = "Normal"
    best_ll = ll_norm
else:
    best_fit = "Exponential"
    best_ll = ll_expon

print("\nBEST FIT")
print("-" * 50)
print(f"Distribution: {best_fit}")
print(f"Log-Likelihood: {best_ll:.2f}")
````
